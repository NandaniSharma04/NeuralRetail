# Revenue Forecasting (Hybrid Proxy)
This notebook trains a time-series model to forecast daily revenue.

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
import joblib
```

```python
np.random.seed(42)
days = np.arange(365)
revenue = 500 + (days * 0.5) + (np.sin(days * (2 * np.pi / 30)) * 100) + np.random.normal(0, 20, 365)
X = pd.DataFrame({'day': days, 'month': (days % 30)})
y = revenue
```

```python
model = GradientBoostingRegressor()
model.fit(X, y)
print(f'R^2 Score: {model.score(X, y):.3f}')
joblib.dump(model, '../models/revenue_forecast_model.pkl')
```
