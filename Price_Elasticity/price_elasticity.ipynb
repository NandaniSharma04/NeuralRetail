# Price Elasticity Modeling
This notebook trains a Log-Log OLS regression model to estimate price elasticity of demand.

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import joblib
```

```python
np.random.seed(42)
prices = np.random.uniform(10, 100, 1000)
comp_prices = prices + np.random.normal(0, 5, 1000)
ad_spend = np.random.uniform(100, 1000, 1000)
log_demand = 10 - 1.5 * np.log(prices) + 0.5 * np.log(comp_prices) + 0.2 * np.log(ad_spend)
X = pd.DataFrame({'log_price': np.log(prices), 'log_comp_price': np.log(comp_prices), 'log_ad': np.log(ad_spend)})
y = log_demand
```

```python
model = LinearRegression()
model.fit(X, y)
print(f'Price Elasticity: {model.coef_[0]:.2f}')
joblib.dump(model, '../models/elasticity_model.pkl')
```
