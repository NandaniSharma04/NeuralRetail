# Inventory Optimization (EOQ)
This notebook defines the Economic Order Quantity calculation model.

```python
import numpy as np
import joblib

class EOQModel:
    def predict(self, demand_rate, order_cost, holding_cost):
        return np.sqrt((2 * demand_rate * order_cost) / holding_cost)
```

```python
model = EOQModel()
print(f'EOQ for D=1000, S=50, H=2: {model.predict(1000, 50, 2):.2f}')
joblib.dump(model, '../models/eoq_calculator.pkl')
```
